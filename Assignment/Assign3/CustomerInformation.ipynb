```python
# Customer Name Processing (Strings)

name = input("Enter Customer Name: ")

# Remove leading and trailing spaces
name = name.strip()

# 1. Uppercase
print(name.upper())

# 2. Lowercase
print(name.lower())

# 3. Cleaned name
print(name)

# 4. Count the number of characters
print("Length:", len(name))

# 5. Check whether Smith exists
if "smith" in name.lower():
    print("Smith exists")
else:
    print("Smith does not exist")
```

```python
#Unique Customer Cities (Sets)
cities = {"Chennai", "Coimbatore", "Madurai", "Chennai"}

#Display all unique cities.
print(set(cities))


#Count unique cities.
print(len(cities))

#Add a new city.
cities.add("Gobi")
print(cities)

#Remove a city.
cities.remove("Chennai")
print(cities)

#Check whether a city exists.
print("Gobi" in cities)
```

```python
#Dictionaries:
customers = [
    {
        "Customer_ID": "C101",
        "Name": "John",
        "City": "Chennai",
        "Satisfaction_Rating": 4.5
    },
    {
        "Customer_ID": "C102",
        "Name": "Mary",
        "City": "Coimbatore",
        "Satisfaction_Rating": 4.5
    },
    {
        "Customer_ID": "C103",
        "Name": "David",
        "City": "Madurai",
        "Satisfaction_Rating": 4.2
    }
]


#Display all customer details:
print(customers)

#Update satisfaction rating.
for customer in customers:
    if customer["Customer_ID"] == "C102":
        customer["Satisfaction_Rating"] = 4.8


#Add a new field:Membership Type
customers[0]["Membership"]="Gold"
customers[1]["Membership"]="Silver"
customers[2]["Membership"]="Gold"
print(customers)

#Remove a field.
for customer in customers:
    del customer["Membership"]
print(customers)

#Display all keys and values separately.
for customer in customers:
    print("Keys:", customer.keys())
    print("Values:", customer.values())
```
